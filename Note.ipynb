**Теперь надо выделить параметр месяца рождения**

```python
years = list(laureates['birthDate'].values)
years = [i for i in years if type(i) == str]
for i in range(len(years)):
    years[i] = int(str(years[i])[5:7])
years
laureates['mounth'] = pd.Series(years)
```

**Также надо учитывать то что месяц может быть не указан (из-за неуказанной даты), либо равен 0, обработаем такие случаи**

```python
laureates = laureates[laureates['mounth'].notna()]
laureates = laureates[laureates['mounth'] != 0]
```

```python
laureates
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
```

```python
num_cols = ['ageAtAward']
cat_cols = ['gender', 'portion', 'mounth', 'birthCountry']
```

```python
X = laureates[num_cols + cat_cols]
y = laureates['category']
```

**Для числовых параметров применим масштабирование, для того чтобы применять категориальные переменные применим OneHotEncoder**

```python
pre = ColumnTransformer(
    transformers = [
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(handle_unknown='ignore'), cat_cols)
    ],
    remainder = 'drop'
)
```

**настройка Pipeline**

```python
pipe = Pipeline([
    ('pre', pre),
    ('log', LogisticRegression())
])
```

**ну и пришло время разделить данные на тестовую и тренировочные выборки**

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=0, stratify=y
)
```

```python
X_train
```

```python
y_train
```

```python
pipe.fit(X_train, y_train)
```

**проверим нашу модель на Дональде Трампе**

```python
new_frame = pd.DataFrame(
    [[80, 'male', '1', 6, 'USA']],
    columns=['ageAtAward', 'gender', 'portion', 'mounth', 'birthCountry']
)
```

**Ну и в общем пора проверять ошибку модели((**

```python
from sklearn.metrics import log_loss
```

```python
y_pred_proba = pipe.predict_proba(X_test)
log_loss(y_test, y_pred_proba)
```
